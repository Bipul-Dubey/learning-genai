```python
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import numpy as np 
```

```python
## load the ANN trained model, scaler pickle file, onehotencoding

model=load_model("model.h5")

## load pickle file - label encoder geography
with open("onehot_encoder_geo.pkl", "rb") as file:
    label_encoder_geo=pickle.load(file)

# label encoder gender
with open("label_encoder_gender.pkl","rb") as file:
    label_encoder_gender=pickle.load(file)

## load scaler pickle file
with open("scaler.pkl","rb") as file:
    scaler=pickle.load(file)
```

```python
## example data
input_data={
    "CreditScore": 750,
    "Geography": "France",
    "Gender": "Male",
    "Age": 40,
    "Tenure": 3,
    "Balance": 60000,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 50000
}
```

```python
# onehot encoding of geography
geo_encoded=label_encoder_geo.transform([[input_data["Geography"]]]).toarray()
geo_endcoded_df=pd.DataFrame(geo_encoded,columns=label_encoder_geo.get_feature_names_out(["Geography"]))
geo_endcoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## encode categorical variable - gender
input_df["Gender"]=label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
## concatination with OHE data
input_df=pd.concat([input_df.drop("Geography",axis=1), geo_endcoded_df],axis=1)
input_df
```

```python
## scaling the input data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## prediction churn
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_probability=prediction[0][0]
print(prediction_probability)

if prediction_probability > 0.5:
    print("The customer is likely to move.")
else:
    print("The customer is not likely to move.")
```
