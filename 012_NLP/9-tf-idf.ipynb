# TF-IDF -> Term Frequency - Inverse Document Frequency

```python
# load the sample dataset
import pandas as pd
messages = pd.read_csv("../sampledataset/spam.csv", encoding="latin-1")
```

```python
messages
```

```python
# data cleaning and pre processing
import re
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

wnl=WordNetLemmatizer()
```

```python
corpus = []
# loop
for i in range(len(messages)):
 # remove symbols keep only words/letter
 review = re.sub("[^a-zA-Z]", " ", messages["message"][i])
 # lower
 review = review.lower()
 # split into chars
 review = review.split()
 # stop words
 review = [wnl.lemmatize(word) for word in review if not word in stopwords.words("english")]
 review = " ".join(review)
 corpus.append(review)
 
```

```python
corpus
```

## Create TF-IDF And NGrams

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=100)
X=tfidf.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: "%.3g" % x))
```

```python
tfidf = TfidfVectorizer(max_features=100, ngram_range=(2,2))
X=tfidf.fit_transform(corpus).toarray()
```
