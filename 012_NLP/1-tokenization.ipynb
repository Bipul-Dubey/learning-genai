```python
import nltk
nltk.download('punkt_tab')
```

```python
corpus = """Hello, this is Bipul Dubey. I'm Software Engineer! Learning Generative AI."""
```

```python
documents=nltk.sent_tokenize(corpus)
print(documents)
```

```python
documents=nltk.word_tokenize(corpus)
print(documents)
```

```python
documents=nltk.wordpunct_tokenize(corpus)
print(documents)
```

```python
tokenizer=nltk.TreebankWordTokenizer()
tokenizer.tokenize(corpus) # the . in this is treated as seperate word when it is last of paragraph
```
