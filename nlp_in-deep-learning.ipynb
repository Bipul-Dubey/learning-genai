# Recurrent Neural Network (RNN): Forward Propagation Through Time

## 1. Architecture: Forward Propagation Through Time

Forward propagation passes an input sequence through a Recurrent Neural Network (RNN) one time step at a time. At each time step, the network combines the current input with the previous hidden state to calculate the new hidden state and, optionally, an output.

### Diagram

```text
                 Forward Propagation →

          t = 1         t = 2         t = 3         t = 4

            b_h           b_h           b_h           b_h
             ↓             ↓             ↓             ↓

          ┌──────┐      ┌──────┐      ┌──────┐      ┌──────┐
x_1 ─────→ │  h_1 │ ───→ │  h_2 │ ───→ │  h_3 │ ───→ │  h_4 │
          └──────┘      └──────┘      └──────┘      └──────┘
                            ↑             ↑             ↑
                           x_2           x_3           x_4

             ↓             ↓             ↓             ↓
            o_1           o_2           o_3           o_4
```

Each hidden state depends on the current input and the previous hidden state. The initial hidden state \(h_0\) is usually initialized to zeros or learned as a parameter.
