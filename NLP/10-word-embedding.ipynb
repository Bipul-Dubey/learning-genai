**Word2Vec** is a popular natural language processing technique that turns words into numerical vectors to help computers understand word meanings and relationships

"Word2Vec uses neural network model to learn word associations from large corpus of text. Once trained, such a model can detect synonymous words or suggests additional word for a partial sentence. As name implies, word2vec represents each distinct word with a particular list of number called vector"

**Continuous bag-of-words model:** predicts the middle word based on surrounding context words. The context consists of a few words before and after the current (middle) word. This architecture is called a bag-of-words model as the order of words in the context is not important.

**Continuous skip-gram model:** predicts words within a certain range before and after the current word in the same sentence. A worked example of this is given below.

advantages of Word2Vec
- sparse matrix -> dense matrix
- semantic meaning is getting captured (vector are same for similar words/sentence)
- fixed vocabulary sizes - fixed set of dimension - vectors
