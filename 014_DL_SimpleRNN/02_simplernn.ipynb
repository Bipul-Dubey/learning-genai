## E2E Deep Leaning Project Using Deep Learning

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense, Input
```

```python
# load the imdb dataset

max_features = 10000 # vocabulary size
(X_train, y_train),(X_test, y_test)= imdb.load_data(num_words=max_features)

# print the shape of data
print(f'Training data shape {X_train.shape}, Training label shape: {y_train.shape}')
print(f'Testing data shape {X_test.shape}, Testing label shape: {y_test.shape}')
```

```python
# inspect sample review and its label
sample_review = X_train[0]
sample_label = y_train[0]

print(f'Sample review as int: {sample_review}')
print(f'sample label: {sample_label}')
```

```python
## mapping of words index back to words for understanding
word_index = imdb.get_word_index()
# word_index
## reverse word index
reverse_word_index = {value: key for key, value in word_index.items()}
reverse_word_index
```

```python
# decode the review
decoded_review = ' '.join([reverse_word_index.get(i - 3, "?") for i in sample_review])
decoded_review
```

```python
max_len = 500

X_train = sequence.pad_sequences(X_train, maxlen=max_len)
X_test = sequence.pad_sequences(X_test, maxlen=max_len)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
## Train our simple RNN
model = Sequential([
    Input(shape=(max_len,)),
    Embedding(input_dim=max_features, output_dim=128, input_length=max_len),
    # SimpleRNN(128, activation="relu"),
    SimpleRNN(128),
    Dense(1, activation="sigmoid")
])
```

```python
model.summary()
```

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
# creating an instance of EarlyStopping callback
from tensorflow.keras.callbacks import EarlyStopping
earlystopping=EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
earlystopping
```

```python
# train the model with early stopping
history = model.fit(
    X_train,y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    callbacks=[earlystopping]
)
```

```python
# save the model
model.save("simple_rnn_imdb.h5")
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print("Test loss:", test_loss)
print("Test accuracy:", test_accuracy)
```
